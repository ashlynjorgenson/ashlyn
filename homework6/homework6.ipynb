```python
# 4: Prime Star Systems
arr=np.array([
    [2,3,5],
    [14,6,8],
    [11,13,17],
    [7,10,13],
])
def is_prime(n):
    if n<2:
        return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True
def prime_rows(array):
    ndarray=[]
    for row in array:
        if any(is_prime(n) for n in row):
            ndarray.append(row)
    return np.array(ndarray)
result=prime_rows(arr)
print(result)
```

```python
def odd_number(rows, cols):
    array=np.zeros((rows,cols))
    for row in range(rows):
        if row % 2 == 1:
            for col in range(cols):
                if col % 2 ==1:
                    array[row, col]=1
        else:
            for col in range(cols):
                if col % 2 == 0:
                    array[row,col]=1
    return array
result=odd_number(8, 8)
print(result)
```

```python
reverse= 1 - result
print(reverse)
```

```python
def expanding_universe(str, int):
    spaces= ' ' * int
    result = spaces.join(str)
    return result
print(expanding_universe("Galaxy", 2))
```
