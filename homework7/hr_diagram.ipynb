```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data=np.loadtxt("../../course_assignments/homework7/star_data.txt",
                skiprows=1)
print(data.shape)

magnitude = data[:, 0]
temperature = data[:, 1]

min_mag=np.min(magnitude)
max_mag=np.max(magnitude)
brightest_star=min_mag
print("Brightest Star")
print(brightest_star)

dimmest_star=max_mag
print("Dimmest Star")
print(dimmest_star)


min_temp=np.min(temperature)
max_temp=np.max(temperature)

hottest_star=max_temp
print("Hottest Star")
print(hottest_star)

coolest_star=min_temp
print("Coolest Star")
print(coolest_star)

log_magnitude=np.log(magnitude)
log_temperature=np.log(temperature)

plt.figure(figsize=(8,6))
plt.scatter(log_temperature, log_magnitude, label="Stars", s=20, color="darkgoldenrod")

brightest=np.where(magnitude==brightest_star)[0]
plt.scatter(log_temperature[brightest], log_magnitude[brightest],
            color="black", marker="*", s=80, label="Brightest Star!")

dimmest=np.where(magnitude==dimmest_star)[0]
plt.scatter(log_temperature[dimmest], log_magnitude[dimmest],
            color="purple", marker="*", s=80, label="Dimmest Star :( ")

hottest=np.where(temperature==hottest_star)[0]
plt.scatter(log_temperature[hottest], log_magnitude[hottest],
            color="green", marker="*", s=80, label="Hottest Star :o")

coolest=np.where(temperature==coolest_star)[0]
plt.scatter(log_temperature[coolest], log_magnitude[coolest],
            color="darkorange", marker="*", s=80, label="Coolest Star ;)")

plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")
plt.title("Hertzsprung-Russel Diagram")

plt.legend()

plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

plt.show()

from plt import colormaps
list(colormaps)
```
