```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x, A, w):
    return A*np.sin(w*x)
    
x=np.linspace(0, 2*np.pi, 1000)
fig, axes=plt.subplots(1,3, figsize=(25,5))

amplitudes=[0.5,1,1.5,2,2.5]
for A in amplitudes:
    axes[0].plot(x, make_sine_wave(x, A, 1), label=f'A{A}, w=1')
axes[0].set_title("Varying Amplitude")
axes[0].set_xlabel("Input Array")
axes[0].set_ylabel("y(x)=A*sin(wx)")
axes[0].legend()

frequencies=[1,2,3,4,5]
for w in frequencies:
    axes[1].plot(x, make_sine_wave(x, 1, w), label=f'A=1, w{w}')
axes[1].set_title("Varying Frequency")
axes[1].set_title("Varying Amplitude")
axes[1].set_xlabel("Input Array")
axes[1].set_ylabel("y(x)=A*sin(wx)")
axes[1].legend()

for i, (A, w) in enumerate(zip([0.5, 1, 1.5, 2, 2.5], [1,2,3,4,5])):
    axes[2].plot(x, make_sine_wave(x, A, w), label=f'A={A}, w={w}')
axes[2].set_title("Varying Frequency and Amplitude")
axes[2].set_title("Varying Amplitude")
axes[2].set_xlabel("Input Array")
axes[2].set_ylabel("y(x)=A*sin(wx)")
axes[2].legend()
plt.show()

```
