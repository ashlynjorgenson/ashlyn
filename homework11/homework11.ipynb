```python
import numpy as np
import matplotlib.pyplot as plt

# Load image data (no headers)
data = np.loadtxt("astro_image.csv")

plt.figure(figsize=(8, 6))
img = plt.imshow(data, origin="lower")
plt.colorbar(img, label="Pixel Intensity")
plt.title("Full Astro Image")
plt.xlabel("X Pixel")
plt.ylabel("Y Pixel")
plt.show()
```
