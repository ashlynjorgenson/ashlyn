```python
import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits
```

```python
fits_files = ["M13_0001.fits", "M13_0002.fits", "M13_0003.fits", "M13_0004.fits", "M13_0005.fits"]

folder_path = "../../course_assignments/homework10/"
data_list=[]
for filename in fits_files:
    file_path = folder_path + filename
    with fits.open(file_path) as hdul:
        header = hdul[0].header
        data = hdul[0].data
        data_list.append(data)
data_list
```

```python
fig, axes = plt.subplots(1, 5, figsize=(20, 4)) 

for i, data in enumerate(data_list):
    axes[i].imshow(data, origin="lower") 
    axes[i].set_title(f"FITS {i+1}")      
    axes[i].axis("off")                    

plt.suptitle("FITS Images (Individual Scales)")
plt.show()
```

```python
vmin = min([d.min() for d in data_list])
vmax = max([d.max() for d in data_list])

fig, axes = plt.subplots(1, 5, figsize=(20, 4))
for i, data in enumerate(data_list):
    im = axes[i].imshow(data, origin="lower", vmin=vmin, vmax=vmax)
    axes[i].set_title(f"FITS {i+1}")
    axes[i].axis("off")

cbar = fig.colorbar(im, ax=axes.ravel().tolist(), shrink=0.8)
plt.suptitle("FITS images (same color scale)")
plt.show()
```

```python
# the color is the same across the FITS images
```

```python
for i, data in enumerate(data_list):
    plt.figure(figsize=(6,4))
    plt.hist(data.ravel(), bins=50, color='skyblue', edgecolor='black') 
    plt.title(f"Histogram of FITS {i+1} Pixel Values")
    plt.xlabel("Pixel Intensity")
    plt.ylabel("Number of Pixels")
    plt.show()
# x axis = pixel intensity
# y axis = number of pixels
# bins=50 means to divide the range of pizels into 50 intervals
# centered around 500
# not very spread out. very concentrated at 500
```

```python
for i in range(len(data_list)):
    pixels = data_list[i].ravel()       
    mean_val = np.mean(pixels)         
    median_val = np.median(pixels)     
    print("FITS", i+1, "Mean =", round(mean_val, 2), "Median =", round(median_val, 2))
```
