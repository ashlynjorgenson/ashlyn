```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from astropy import units as u
from astropy import constants as const
```

```python
df=pd.read_csv("../../course_assignments/homework9/solar_system.csv")
print(df.shape)
```

```python
df2 = df.set_index("Attribute").T # makes 'attribute' a row name .T makes rows columns and columns rows
print(df2.shape)
df2.index.name = "Planet" # names the index planet
print(df2.shape)
df2.reset_index(inplace=True) # moves the index planet to the columns with the planets
print(df2.shape)
df2.columns.name = None # removes Attriute index name
```

```python
skip = ["Planet", "Ring System?", "Global Magnetic Field?"]

for col in df2.columns:
    if col not in skip:
        df2[col] = pd.to_numeric(df2[col], errors="coerce")

print(df2.dtypes)

print(df2.head())
```

```python
# object, float64, int64
# float64 = number with decimals
# object generic for anything that isn't a number
# whole number
# Planet, ring system, and glob magnetic field because we didnt include them because they have no units
```

```python
def attach_units(df, old_col, unit, new_col=None):
    df[old_col] = [val * unit for val in df[old_col]]
    if new_col:
        df.rename(columns={old_col: new_col}, inplace=True)

attach_units(df2, "Length of Day (hours)", u.hour)

print(df2["Length of Day (hours)"])
```

```python
# print(df2.columns.tolist())
```

```python
semi_major_axis = (df2["Perihelion (10^6 km)"] + df2["Aphelion (10^6 km)"]) / 2 * 1e6 * u.km

if "Semi-Major Axis (km)" in df2.columns:
    df2["Semi-Major Axis (km)"] = semi_major_axis
else:
    aphelion_index = df2.columns.get_loc("Aphelion (10^6 km)")



print(df2[["Planet", "Semi-Major Axis (km)"]])
```

```python
df2["Orbital Period (years)"] = df2["Orbital Period (days)"] * u.day.to(u.year)  # multiply by conversion factor

print(df2["Orbital Period (years)"])

earth_period = df2.loc[df2["Planet"] == "Mars", "Orbital Period (years)"].values[0]
print(f"Mars's orbital period: {earth_period:.4f} years")
```

```python
print(const.au)

print(f"1 AU = {const.au.to(u.km):.2f} km")
distance_cols = [
    "Diameter (km)",
    "Distance from Sun (10^6 km)",
    "Perihelion (10^6 km)",
    "Aphelion (10^6 km)",
    "Semi-Major Axis (km)"
]

for col in distance_cols:
    new_col = col.replace("(km)", "(AU)").replace("(10^6 km)", "(AU)")
    df2[new_col] = (df2[col] * u.km) / const.au  # convert to AU


favorite = "Mars"
for col in distance_cols:
    new_col = col.replace("(km)", "(AU)").replace("(10^6 km)", "(AU)")
    val = df2.loc[df2["Planet"] == favorite, new_col].values[0]
    print(f"{new_col} for {favorite}: {val}")
```
