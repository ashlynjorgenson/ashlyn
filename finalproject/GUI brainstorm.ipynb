```python
def button_callback():
    print("Button clicked!")

def main():
    dpg.create_context()
    
    with dpg.window(label="Interactive GUI"):
        dpg.add_text("Click the button below:")
        dpg.add_button(label="Click Me!", callback=button_callback)
    
    dpg.create_viewport(title='Interactive Example', width=400, height=200)
    dpg.setup_dearpygui()
    dpg.show_viewport()
    dpg.start_dearpygui()
    dpg.destroy_context()

if __name__ == "__main__":
    main()
```

```python
# main script must always be:
# dpg.create_context()
# dpg.create_viewport(includes title, w, h)
# dpg.setup_dearpygui()
# dpg.show_viewport()
# dpg.start_dearpygui()
# dpg.destroy_context()

# Items, UI Items, Containers
# Items: anything in the library (buttons, registries, windows)
# UI Items: Any item in DPG with a visual component (button, listbox, window)
# Containers: items that hold other items (window, groups, registries)

# The Primary Window
# One window can be assigned as the primary
# Fills the viewport
# Items are created by add_
# All items must have a tag which is a integer or string that refer to the item after it has been created. Items return their tag when they are created
```

```python
import dearpygui.dearpygui as dpg

dpg.create_context()

with dpg.window(tag="Primary Window"):
    dpg.add_text("Enter a Planet Below")
    dpg.add_input_text()
    dpg.add_button(label="Submit")

dpg.create_viewport(title='Habitability Calculator', width=600, height=200)
dpg.setup_dearpygui()
dpg.show_viewport()
dpg.set_primary_window("Primary Window", True)
dpg.start_dearpygui()
dpg.destroy_context()
```

```python




def submit_callback():
    planet_name = dpg.get_value("planet_input")
    print(f"Planet submitted: {planet_name}")
    dpg.set_value("output_text", f"Habitability for {planet_name}: ??%")

dpg.create_context()

width, height, channels, data = dpg.load_image("Untitled_Artwork.png")
with dpg.texture_registry():
    dpg.add_static_texture(width, height, data, tag="image_id")

with dpg.window(no_title_bar=True, no_resize=True, no_move=True):
    with dpg.drawlist(width=390, height=844):
        dpg.draw_image("image_id", (0, 0), (390, 844), uv_min=(0, 0), uv_max=(1, 1))

with dpg.window(label="Input Window", pos=(400, 50), width=300, height=150):
    dpg.add_text("Enter a Planet Name:")
    dpg.add_input_text(tag="planet_input", width=200)
    dpg.add_button(label="Submit", callback=submit_callback)
    dpg.add_text("", tag="output_text")  # Placeholder for output

dpg.create_viewport(title="Habitability Calculator", width=720, height=900)
dpg.setup_dearpygui()
dpg.show_viewport()
dpg.start_dearpygui()
dpg.destroy_context()
```
